'Public Ivies" Sub-Data Frame

```python
schools = [
    'University of Virginia',
    'College of William & Mary',
    'University of Texas at Austin',
    'University of California, Los Angeles',
    'University of North Carolina - Chapel Hill'
]

peers_data = df[df['Institution Name'].isin(schools)]
peers_data.groupby('Institution Name')['Foreign Gift Amount'].agg(['count', 'sum', 'mean', 'median'])
```

Total Foreign Gift Amount by Institutin Name

```python
sns.barplot(data=peers_data, x='Foreign Gift Amount', y='Institution Name',estimator='sum', errorbar=None)
plt.xlabel('Total Foreign Gift Amount')
plt.ylabel('Institution Name')
plt.title('Total Foreign Gift Amount by Institution Name')
plt.show()
```

Total Number of Gift Recieved

```python
sns.countplot(data=peers_data, y='Institution Name')
plt.xlabel('Total Number of Gifts Recieved')
plt.ylabel('Institution Name')
plt.title('Total Number of Gift Recieved by Institution Name')
plt.show()
```

Country of Giftor Diagram

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()


giftor = 'Country of Giftor'
recipi = 'Institution Name'
flow = 'Foreign Gift Amount'
N = 25

flows = (
    peers_data.groupby([giftor, recipi])[flow].sum()
    .reset_index()
    .sort_values(flow, ascending=False)
    .groupby(recipi).head(6)
)

labels = (
    flows[giftor].tolist()
    + flows[recipi].tolist()
)

labels = list(dict.fromkeys(labels))

fig = go.Figure(
    go.Sankey(
        node=dict(label=labels),
        link=dict(
            source=flows[giftor]
                        .map(labels.index),
            target=flows[recipi]
                        .map(labels.index),
            value=flows[flow]
        )
    )
)

fig.show()
```

UVA vs. Overall Market

```python
# total gifts amount
institution_totals = df.groupby('Institution Name')['Foreign Gift Amount'].sum()
uva_total = institution_totals['University of Virginia']
Market_mean = institution_totals.mean()
uva_vs_market = uva_total / Market_mean

# number of donors
uva_donors = uva_data['Giftor Name'].nunique()
avg_donors = df.groupby('Institution Name')['Giftor Name'].nunique().mean()
uva_vs_avg_donors = uva_donors / avg_donors

# number of gifts
institution_counts = df.groupby('Institution Name')['Foreign Gift Amount'].count()
uva_gifts = institution_counts['University of Virginia']
avg_gifts = institution_counts.mean()
uva_vs_avg_gifts = uva_gifts / avg_gifts




print(f"University of Virginia total gifts amount: {uva_total}")
print(f"Market mean gifts: {Market_mean}")
print(f"UVA vs Market ratio: {uva_vs_market}")
print(f"University of Virginia number of donors: {uva_donors}")
print(f"Market mean number of donors: {avg_donors}")
print(f"UVA vs Market donors ratio: {uva_vs_avg_donors}")
print(f"University of Virginia number of gifts: {uva_gifts}")
print(f"Market mean number of gifts: {avg_gifts}")
print(f"UVA vs Market gifts ratio: {uva_vs_avg_gifts}")
```
